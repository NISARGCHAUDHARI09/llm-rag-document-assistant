```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader

```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data" / "raw"

print("Project Root:", PROJECT_ROOT)
print("Data Directory:", DATA_DIR)
```

```python
print("Data directory exists:", DATA_DIR.exists())
```

```python
pdf_files = list(DATA_DIR.glob("*.pdf"))

print(f"Number of PDF files found: {len(pdf_files)}")

for pdf in pdf_files:
    print(pdf.name)
```

```python
pdf_path = pdf_files[0]

loader = PyPDFLoader(str(pdf_path))

documents = loader.load()

print("PDF:", pdf_path.name)
print("Number of pages:", len(documents))
```

```python
all_documents = []

for pdf_path in pdf_files:
    print(f"Loading: {pdf_path.name}")

    loader = PyPDFLoader(str(pdf_path))
    documents = loader.load()

    all_documents.extend(documents)

    print(f"  Pages loaded: {len(documents)}")

print("\nTotal pages loaded:", len(all_documents))
```

```python
print("Total document pages:", len(all_documents))

print("\nFirst document metadata:")
print(all_documents[0].metadata)

print("\nFirst 500 characters:")
print(all_documents[0].page_content[:500])
```

```python
from collections import Counter

page_counts = Counter(
    document.metadata.get("source")
    for document in all_documents
)

for source, count in page_counts.items():
    print(f"{Path(source).name}: {count} pages")
```
