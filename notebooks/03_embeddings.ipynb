```python
import math


def cosine_similarity(vector_a, vector_b):
    """
    Calculate cosine similarity between two vectors.
    """

    dot_product = sum(
        a * b for a, b in zip(vector_a, vector_b)
    )

    magnitude_a = math.sqrt(
        sum(a * a for a in vector_a)
    )

    magnitude_b = math.sqrt(
        sum(b * b for b in vector_b)
    )

    return dot_product / (magnitude_a * magnitude_b)


similarity_1 = cosine_similarity(
    test_embeddings[0],
    test_embeddings[1],
)

similarity_2 = cosine_similarity(
    test_embeddings[0],
    test_embeddings[2],
)

print(
    "Similarity between environmental texts:",
    similarity_1,
)

print(
    "Similarity between environmental and weather text:",
    similarity_2,
)
```

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
from src.document_loader import load_pdfs_from_directory
from src.text_splitter import split_documents
from src.embeddings import get_embedding_model

print("Imports successful.")
```

```python
DATA_DIR = PROJECT_ROOT / "data" / "raw"

documents = load_pdfs_from_directory(DATA_DIR)

print("Total pages loaded:", len(documents))
```

```python
chunks = split_documents(documents)

print("Original pages:", len(documents))
print("Generated chunks:", len(chunks))
```

```python
embedding_model = get_embedding_model()

print("Embedding model loaded successfully.")
```

```python
sample_chunks = chunks[:5]

print("Number of sample chunks:", len(sample_chunks))

for i, chunk in enumerate(sample_chunks):
    print("=" * 80)
    print(f"CHUNK {i}")
    print("Source:", chunk.metadata["source_file"])
    print("Page:", chunk.metadata["page_label"])
    print("Chunk ID:", chunk.metadata["chunk_id"])
    print("Text preview:")
    print(chunk.page_content[:500])
```

```python
sample_texts = [
    chunk.page_content
    for chunk in sample_chunks
]

sample_embeddings = embedding_model.embed_documents(
    sample_texts
)

print("Number of embeddings:", len(sample_embeddings))
print("Embedding dimension:", len(sample_embeddings[0]))
```

```python
all_texts = [
    chunk.page_content
    for chunk in chunks
]

print("Total texts to embed:", len(all_texts))
```

```python
print("Starting full dataset embedding...")

all_embeddings = embedding_model.embed_documents(
    all_texts
)

print("Embedding completed.")
print("Total embeddings:", len(all_embeddings))
print("Embedding dimension:", len(all_embeddings[0]))
```
