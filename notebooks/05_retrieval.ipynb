```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
from src.embeddings import get_embedding_model
from src.vector_store import load_vector_store
from src.retriever import (
    create_retriever,
    retrieve_documents,
    format_retrieved_documents,
)

from config.settings import TOP_K
```

```python
embedding_model = get_embedding_model()

print("Embedding model loaded successfully.")
```

```python
vector_store = load_vector_store(
    embedding_model=embedding_model
)

print("ChromaDB loaded successfully.")
print("Documents stored:", vector_store._collection.count())
```

```python
retriever = create_retriever(
    vector_store=vector_store,
    top_k=TOP_K,
)

print("Retriever created successfully.")
print("Top K:", TOP_K)
```

```python
query = (
    "What are the environmental impacts "
    "of oil and gas development?"
)

retrieved_documents = retrieve_documents(
    retriever=retriever,
    query=query,
)

print("Query:")
print(query)

print("\nNumber of retrieved documents:")
print(len(retrieved_documents))
```

```python
for i, document in enumerate(
    retrieved_documents,
    start=1,
):
    print("=" * 80)
    print(f"RESULT {i}")
    print("=" * 80)

    print("Source:", document.metadata.get(
        "source_file",
        "Unknown"
    ))

    print("Page:", document.metadata.get(
        "page",
        "Unknown"
    ))

    print("\nContent:")
    print(document.page_content[:1000])
    print()
```

```python
query_2 = (
    "What wildlife and habitat effects "
    "could result from the proposed project?"
)

retrieved_documents_2 = retrieve_documents(
    retriever=retriever,
    query=query_2,
)

print("Query:")
print(query_2)

print("\nRetrieved sources:")

for i, document in enumerate(
    retrieved_documents_2,
    start=1,
):
    print(
        f"{i}. "
        f"{document.metadata.get('source_file', 'Unknown')} "
        f"| Page: "
        f"{document.metadata.get('page', 'Unknown')}"
    )
```

```python
formatted_results = format_retrieved_documents(
    retrieved_documents
)

for i, (
    source_file,
    page,
    content,
) in enumerate(
    formatted_results,
    start=1,
):
    print("=" * 80)
    print(f"RESULT {i}")
    print("=" * 80)

    print("Source:", source_file)
    print("Page:", page)
    print("Content length:", len(content))
    print()
```

```python
assert len(retrieved_documents) == TOP_K

for document in retrieved_documents:
    assert document.page_content.strip()
    assert document.metadata.get("source_file") is not None
    assert document.metadata.get("page") is not None

print("Retrieval validation passed.")
```
