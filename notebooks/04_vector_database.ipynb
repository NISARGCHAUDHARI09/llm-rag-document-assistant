```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
from src.document_loader import load_pdfs_from_directory
from src.text_splitter import split_documents
from src.embeddings import get_embedding_model
from src.vector_store import create_vector_store

print("Imports successful.")
```

```python
DATA_DIR = PROJECT_ROOT / "data" / "raw"

documents = load_pdfs_from_directory(DATA_DIR)

print("Total pages loaded:", len(documents))
```

```python
chunks = split_documents(documents)

print("Generated chunks:", len(chunks))
```

```python
embedding_model = get_embedding_model()

print("Embedding model loaded successfully.")
```

```python
print("Creating ChromaDB vector store...")
print("This may take some time because the documents need to be embedded.")

vector_store = create_vector_store(
    documents=chunks,
    embedding_model=embedding_model,
)

print("ChromaDB vector store created successfully.")
```

```python
print("Vector store type:", type(vector_store))

print(
    "Documents stored:",
    vector_store._collection.count()
)
```
